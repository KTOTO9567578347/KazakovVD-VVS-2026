# Контрольный вопрос. Почему Pclass, содержащий целые числа, следует рассматривать как категориальный признак?

Потому что это три класса: 1, 2, 3 - это не числовые значения, а категории качества обслуживания пассажиров закодированные цифрами.

```python
na_cnts = df.isna().sum()
na_perc = ((na_cnts/len(df))*100).round(2)

pd.DataFrame(data = {"cnts": na_cnts, "perc": na_perc}, index=na_perc.index)
```

Больше всего пропусков - в столбце Cabin.

В столбце Age - 86 пропусков.

```python
descr_before = df['Age'].describe().loc[DESCR_CHARS]
descr_before['nan_cnt'] = na_cnts['Age']

df['Age'] = df['Age'].fillna(value = df['Age'].median())

descr_after = df['Age'].describe().loc[DESCR_CHARS]
descr_after['nan_cnt'] = df['Age'].isna().sum()

pd.DataFrame(data = {"BEFORE: ": descr_before, "AFTER: ": descr_after}, index=descr_before.index)
```

```python
print("Строк и столбцов до дропа: ", df.shape)

df = df.dropna(subset=['Embarked'])
if na_perc['Cabin'] > 50:
    df = df.drop(labels = "Cabin", axis=1)

print("Строк и столбцов после дропа: ", df.shape)
```

Контрольный вопрос.

И (&) требует соблюдения ВСЕХ перечисленных условий для данной строки, чтобы она была включена в результируюший индекс, а ИЛИ (|) хотя-бы одного.

```python
group_1 = df.groupby(by=["Pclass", "Sex"])[["PassengerId", "Age", "Fare", "Survived"]]
group_1_agg_res = group_1.aggregate(func={"PassengerId": 'count', "Age": 'mean', "Fare": 'mean', "Survived": 'mean'}) 
group_1_agg_res['Survived'] = round(group_1_agg_res['Survived']*100, 2)
group_1_agg_res.to_csv("../res/titanic_statistics.csv")
group_1_agg_res
```

```python
group_2 = df.groupby(by=["Embarked"])[["PassengerId","Survived"]]
group_2_agg_res = group_2.aggregate(func={"PassengerId": 'count', "Survived": 'mean'}) 
group_2_agg_res['Survived'] = round(group_2_agg_res['Survived']*100, 2)
group_2_agg_res
```

```python
group_1 = df.groupby(by=["Pclass"])[["Survived"]]
round(group_1.mean()*100, 2)
```
